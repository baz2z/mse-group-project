```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
import re
```

```python
docs = [
    "The sky is blue.",
    "The sun is bright today.",
    "The sun in the sky is bright.",
    "We can see the shining sun, the bright sun."
]
```

```python
# Preprocessing function
def preprocess(text):
    stemmer = PorterStemmer()
    text = re.sub(r'\W', ' ', text)  # Remove all non-word characters
    text = text.lower()  # Convert to lowercase
    tokens = text.split()  # Tokenize
    tokens = [stemmer.stem(word) for word in tokens if word not in stopwords.words('english')]  # Stemming and stopword removal
    return ' '.join(tokens)
```

```python
# Preprocess documents
preprocessed_docs = [preprocess(doc) for doc in docs]

# Vectorize text
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(preprocessed_docs)

print(X)

# Compute TF-IDF
tfidf_transformer = TfidfTransformer()
tfidf_matrix = tfidf_transformer.fit_transform(X)

# tfidf_matrix is a sparse matrix representation of the TF-IDF values
print(tfidf_matrix)


# Query term
q = "sun"
# Get the index of the query term in the vocabulary
q_index = vectorizer.vocabulary_.get(q)

# Get the tfidf score for the query term in each document
tfidf_scores = tfidf_matrix[:, q_index].toarray().flatten()

# Print the tfidf scores
print(tfidf_scores)
```
